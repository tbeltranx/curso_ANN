Este código genera un **mapa de calor (*heatmap*) para visualizar la distribución de los valores faltantes** en el DataFrame original (`df`), utilizando las bibliotecas Matplotlib y Seaborn.

La representación gráfica permite identificar visualmente qué variables presentan datos faltantes (`NaN`), cómo se distribuyen entre los registros y si existen patrones de ausencia de información.

### Procedimiento

1. **Define las dimensiones de la figura** mediante `plt.figure(figsize=(20, 10))`. Se establece un área gráfica de 20 pulgadas de ancho por 10 pulgadas de alto, proporcionando espacio suficiente para representar las filas y columnas del DataFrame.

   - **`False`:** indica que el elemento contiene un valor no nulo.

3. **Genera el mapa de calor** mediante `sns.heatmap(df.isnull(), cbar=False, cmap='viridis')`. Cada celda de la matriz se representa mediante un color, permitiendo distinguir visualmente los datos disponibles de los faltantes.

4. **Aplica la paleta de colores `viridis`**, que utiliza diferentes tonalidades para representar los valores booleanos:

7. **Muestra la representación gráfica** mediante `plt.show()`, permitiendo observar la distribución de los valores nulos en el conjunto de datos.

### Interpretación de la gráfica

El mapa de calor representa la estructura de disponibilidad de información del DataFrame:

- **Eje horizontal (X):** corresponde a las columnas o variables del conjunto de datos.
- **Eje vertical (Y):** corresponde a los registros o filas del DataFrame.
- **Regiones moradas:** indican posiciones donde existen datos disponibles.
- **Regiones amarillas:** indican posiciones donde existen valores faltantes.

La distribución de colores permite identificar diferentes situaciones:

1. **Columnas predominantemente moradas:** variables con alta disponibilidad de información.
2. **Columnas predominantemente amarillas:** variables con una proporción elevada de valores faltantes.
3. **Columnas completamente amarillas:** variables que no contienen ningún valor válido.
4. **Franjas horizontales amarillas:** registros que presentan valores faltantes en varias variables.
5. **Bloques o patrones de colores:** posibles agrupaciones de registros con características similares de disponibilidad de información.

Preparar un conjunto de datos para su análisis, conservando únicamente los registros con información de **lignina** (`lignin`) y **grasa** (`fat`), eliminando columnas completamente vacías y seleccionando variables según su disponibilidad de datos. ## 1. Explicación paso a paso

### Paso 1. Filtrar los registros con lignina disponible

```python
df_filtered_lignin = df[df['lignin'].notna()]
```

- `df['lignin']` selecciona la columna de lignina.
- `.notna()` identifica los registros cuyo valor no es nulo.
- `df[...]` conserva únicamente las filas que cumplen esa condición.

**Resultado:** todas las filas del DataFrame filtrado tienen un valor no nulo de lignina.

### Paso 2. Eliminar columnas completamente vacías

```python
df_lignin_non_empty_cols = df_filtered_lignin.dropna(axis=1, how='all')
```

- `dropna()` elimina filas o columnas según sus valores faltantes.
- `axis=1` indica que la operación se aplica a las columnas.
- `how='all'` exige que **todos** los valores de una columna sean nulos para eliminarla.

**Resultado:** se eliminan las columnas sin información en el conjunto ya filtrado por lignina.

### Paso 3. Contar los registros disponibles de grasa

```python
fat_non_null_count = df_lignin_non_empty_cols['fat'].count()
```

`count()` cuenta los valores no nulos. Si existen 100 filas y 75 contienen información de grasa, `fat_non_null_count` será igual a **75**.

```python
non_null_counts_lignin = df_lignin_non_empty_cols.count()
```

La operación produce una serie con el número de valores no nulos de cada variable.

### Paso 5. Seleccionar columnas según la disponibilidad de datos

```python
columns_to_keep_refined = [
    col for col, count in non_null_counts_lignin.items()
    if count >= fat_non_null_count
]
```

Esta **comprensión de listas** recorre los pares `(nombre_de_columna, cantidad_de_valores_no_nulos)` y conserva las columnas cuya cantidad de datos disponibles es **mayor o igual** a la cantidad disponible para `fat`.

Por ejemplo, si `fat` contiene 75 valores no nulos, solo se seleccionan automáticamente las columnas con **75 o más** valores no nulos.

```python
if 'nitrogen' not in columns_to_keep_refined:
```

Si `nitrogen` no cumple el criterio de disponibilidad, se añade de todas maneras a la lista de columnas. Esto constituye una **excepción deliberada** al criterio de selección.

**Importante:** agregar el nombre a la lista no crea una columna inexistente en el DataFrame original.

```python
df_intermediate = df_lignin_non_empty_cols[columns_to_keep_refined]
```

Se crea un DataFrame que contiene exclusivamente las columnas seleccionadas en los pasos anteriores.

### Paso 8. Eliminar las filas sin información de grasa

```python
df_final_cleaned = df_intermediate.dropna(subset=['fat'])
```

Se eliminan las filas en las que `fat` es nulo. **No** se eliminan las filas que tengan valores nulos en otras columnas, como `nitrogen`.

```python
print(f"Shape of DataFrame after filtering (including nitrogen and non-null fat): {df_final_cleaned.shape}")
```

````python
################################################################################
#       Detección de valores atípicos mediante Local Outlier Factor (LOF)      #
################################################################################

from sklearn.neighbors import LocalOutlierFactor
import numpy as np

# Select only numeric columns for LOF, excluding the experimental temperature
numeric_cols_for_lof = df_encoded.select_dtypes(
    include=np.number
).columns.drop(
    ['experimental temperature of anaerobic digestion']
)

# Prepare the data for LOF
X = df_encoded[numeric_cols_for_lof]

# Handle potential NaN values by filling with the mean
X = X.fillna(X.mean())

# Initialize and fit the LOF model
# n_neighbors: number of neighbors for local density calculation
# contamination: expected proportion of outliers
lof = LocalOutlierFactor(n_neighbors=20, contamination='auto')

# Fit the model and predict outlier labels
df_encoded['lof_score'] = lof.fit_predict(X)
df_encoded['lof_negative_factor'] = lof.negative_outlier_factor_

# Identify outliers and sort by anomaly severity
outliers_lof = df_encoded[
    df_encoded['lof_score'] == -1
].sort_values(by='lof_negative_factor')

print(f"Number of outliers detected by LOF: {len(outliers_lof)}")
print("Top 10 potential outliers based on LOF score (most negative scores):")
display(outliers_lof.head(10))
````


### Detección de valores atípicos mediante Local Outlier Factor (LOF)

**Objetivo:** Identificar observaciones atípicas (*outliers*) en el DataFrame `df_encoded` mediante el algoritmo **Local Outlier Factor (LOF)**, que compara la densidad local de cada observación con la de sus vecinos.

### 1. Explicación

**Importación de bibliotecas**

- `LocalOutlierFactor`: importa el algoritmo LOF de `scikit-learn`.

- `n_neighbors=20`: considera 20 vecinos para evaluar la densidad local de cada observación.
- `contamination='auto'`: utiliza el umbral automático de LOF para clasificar observaciones atípicas; no estima directamente su proporción.

````python
################################################################################
#          Pair-plots con identificación de valores atípicos (LOF)             #
################################################################################
#
import seaborn as sns
import matplotlib.pyplot as plt

# Select numeric columns, excluding LOF results
df_pairplot = df_encoded.select_dtypes(include=['number']).drop(
    columns=['lof_score', 'lof_negative_factor'],
    errors='ignore'
).copy()

# Add LOF classification for coloring
df_pairplot['lof_score'] = df_encoded['lof_score'].map({
    1: 'Normal',
    -1: 'Outlier'
})

# Generate pair plots
sns.pairplot(
    df_pairplot,
    hue='lof_score',
    diag_kind='hist',
    corner=True,
    plot_kws={'alpha': 0.5, 's': 20},
    palette={'Normal': 'steelblue', 'Outlier': 'red'}
)

plt.show()
````


### Pair-plots con identificación de valores atípicos (LOF)

- `alpha=0.5`: establece transparencia en los puntos.
- `s=20`: define el tamaño de los puntos.
- `palette`: utiliza azul para observaciones normales y rojo para valores atípicos.

**Resultado:** Se obtiene una matriz de gráficas de dispersión donde:

- **Azul:** observaciones normales (`lof_score = 1`).
- **Rojo:** valores atípicos (`lof_score = -1`).

**Observación:** Los valores atípicos se identifican mediante LOF considerando múltiples variables simultáneamente. Por ello, un punto rojo no necesariamente aparecerá aislado en todas las gráficas individuales.
